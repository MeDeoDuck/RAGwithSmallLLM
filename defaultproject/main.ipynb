```python
tokenizer = transformers.AutoTokenizer.from_pretrained("openai-community/gpt2")
tokenizer.deprecation_warnings["Asking-to-pad-a-fast-tokenizer"] = True

additional_special_tokens = {}
if tokenizer.pad_token is None:
    additional_special_tokens["pad_token"] = "<|padding|>"
if tokenizer.eos_token is None:
    additional_special_tokens["eos_token"] = "<|endoftext|>"
if tokenizer.bos_token is None:
    additional_special_tokens["bos_token"] = "<|beginoftext|>"
if additional_special_tokens:
    print(f"Adding special tokens: {additional_special_tokens}")
    tokenizer.add_special_tokens(additional_special_tokens)
tokenizer.padding_side = "left"
tokenizer.model_max_length = MODEL_CONTEXT_LENGTH
```

```python
@dataclasses.dataclass
class TrainingConfig(object):
    # Device Setting
    device: str = "cuda"
    model_dtype:str = "cast_bf16"
    # set "cast_bf16" if device is upper than ampere architecture for best performance unless use "amp_fp16" for mixed precision
    # ampere architecture means nvidia a100, a40, a6000 or rtx 3000 series or higher

    # Training setting
    batch_size: int = 32
    eval_batch_size: int = 32
    gradient_accumulation_steps: int = 2
    num_train_epochs: int = 5
    max_steps: int = None

    # Optimizer setting
    optimizer_type: str = "adamw"
    learning_rate: float = 3e-4
    weight_decay: float = 0.01
    warmup_steps: int = 1000
    max_grad_norm:float = 3.0
    lr_scheduler_type: str = "linear"

    # Trainer setting
    metric_for_best_model: str = "loss"
    metric_greater_is_better: bool = False
    # save_interval: int = 1000
    eval_interval: int = 1000
    logging_interval: int = 10
    save_total_limit: int = 5

    logging_path: str = os.path.join(LOG_PATH, "train")
    output_path: str = os.path.join(LOG_PATH, "output")

    def to_dict(self):
        return {
            "model_dtype": self.model_dtype,
            "batch_size": self.batch_size,
            "eval_batch_size": self.eval_batch_size,
            "gradient_accumulation_steps": self.gradient_accumulation_steps,
            "num_train_epochs": self.num_train_epochs,
            "max_steps": self.max_steps,
            "optimizer_type": self.optimizer_type,
            "learning_rate": self.learning_rate,
            "weight_decay": self.weight_decay,
            "warmup_steps": self.warmup_steps,
            "max_grad_norm": self.max_grad_norm,
            "lr_scheduler_type": self.lr_scheduler_type,
            "metric_for_best_model": self.metric_for_best_model,
            "metric_greater_is_better": self.metric_greater_is_better,
            "eval_interval": self.eval_interval,
            "logging_interval": self.logging_interval,
            "save_total_limit": self.save_total_limit,
        }

    def from_dict(self, config_dict):
        for key, value in config_dict.items():
            if hasattr(self, key):
                setattr(self, key, value)
            else:
                raise KeyError(f"Invalid key: {key} in config_dict")
        self.__post_init__()
```

```python
@torch.no_grad()
def eval_for_summary(model, dataloader, train_config, tokenizer, max_new_tokens=128):
    if train_config.model_dtype == "cast_bf16":
        model = model.to(torch.bfloat16)

    model = model.to(train_config.device)
    model.eval()
    model.compile()

    prediction = []
    answers = []
    for batch in tqdm.auto.tqdm(dataloader, desc="Evaluating", leave=False):
        batch = {k:v.to(train_config.device) for k,v in batch.items()}
        input_ids = batch.pop("input_ids")
        labels = batch.pop("labels")
        attention_mask = batch.pop("attention_mask")

        generation_input_ids = input_ids.clone()
        generation_attention_mask = attention_mask.clone()
        generation_input_ids[labels.ne(-100)] = tokenizer.pad_token_id
        generation_attention_mask[labels.ne(-100)] = 0

        max_length = generation_attention_mask.sum(-1).max().item()
        refit_generation_input_ids = torch.zeros((input_ids.shape[0], max_length), dtype=torch.long, device=input_ids.device)
        refit_generation_attention_mask = torch.zeros((input_ids.shape[0], max_length), dtype=torch.long, device=input_ids.device)
        for i in range(input_ids.shape[0]):
            length = generation_attention_mask[i].sum().item()
            refit_generation_input_ids[i, -length:] = generation_input_ids[i, generation_attention_mask[i].eq(1)]
            refit_generation_attention_mask[i, -length:] = generation_attention_mask[i, generation_attention_mask[i].eq(1)]

        gen_output = model.generate(
            input_ids=refit_generation_input_ids,
            attention_mask=refit_generation_attention_mask,
            max_new_tokens=max_new_tokens,
            return_response_only=True
        )

        labels[labels.eq(-100)] = tokenizer.pad_token_id
        pred = tokenizer.batch_decode(gen_output, skip_special_tokens=True)
        ans = tokenizer.batch_decode(labels, skip_special_tokens=True)

        prediction.extend(pred)
        answers.extend(ans)

    rouge_score = rouge.compute(predictions=prediction, references=answers, use_stemmer=True)
    metrics = {
        "rouge1": rouge_score["rouge1"].item(),
        "rouge2": rouge_score["rouge2"].item(),
        "rougeL": rouge_score["rougeL"].item(),
        "rougeLsum": rouge_score["rougeLsum"].item(),
        "prediction": prediction,
        "answers": answers,
    }

    return metrics

@torch.no_grad()
def eval_for_classification(model, dataloader, train_config):
    if train_config.model_dtype == "cast_bf16":
        model = model.to(torch.bfloat16)

    model = model.to(train_config.device)
    model.eval()
    model.compile()

    prediction = []
    answers = []
    for batch in tqdm.auto.tqdm(dataloader, desc="Evaluating", leave=False):
        batch = {k:v.to(train_config.device) for k,v in batch.items()}
        input_ids = batch.pop("input_ids")
        labels = batch.pop("labels")
        attention_mask = batch.pop("attention_mask")

        logits,_ = model(input_ids=input_ids, attention_mask=attention_mask)
        pred = logits.argmax(-1)

        prediction.extend(pred.tolist())
        answers.extend(labels.tolist())

    prediction = np.array(prediction)
    answers = np.array(answers)

    accuracy = (prediction == answers).sum() / len(answers)

    metrics = {
        "accuracy": accuracy,
        "prediction": prediction,
        "answers": answers,
    }

    return metrics

@torch.no_grad()
def eval_for_rag(model, dataloader, train_config, tokenizer):
    if train_config.model_dtype == "cast_bf16":
        model.model = model.model.to(torch.bfloat16)

    model.model = model.model.to(train_config.device)
    model.model.eval()
    model.model.compile()

    uids = []
    questions = []
    predictions = []
    answers = []
    for batch in tqdm.auto.tqdm(dataloader, desc="Evaluating", leave=False):
        uid = batch['uid']
        question = batch['question']
        answer = batch['answers']

        # retrieval_augmented_generate
        extra_kw_args = {
            "pad_token_id": tokenizer.eos_token_id,
        } if isinstance(model.model, transformers.generation.GenerationMixin) else {}
        outputs = model.retrieval_augmented_generate(
            queries=question,
            qids=uid,
            max_new_tokens=10,
            k=5,
            **extra_kw_args
        )
        pred = model.tokenizer.batch_decode(outputs, skip_special_tokens=True)

        uids.extend(uid)
        questions.extend(question)
        predictions.extend(pred)
        answers.extend(answer)

    # fill here
    #######
    # predictions = [pred.split("\n")[0] for pred in predictions]
    predictions = [pred.strip().split("\n")[0].strip() for pred in predictions]
    #######


    accuracy = best_subspan_exact_match(predictions, answers)
    rouge_score = rouge.compute(predictions=predictions, references=answers)

    metrics = {
        "accuracy": accuracy['acc'],
        "rouge1": rouge_score["rouge1"].item(),
        "rouge2": rouge_score["rouge2"].item(),
        "rougeL": rouge_score["rougeL"].item(),
        "rougeLsum": rouge_score["rougeLsum"].item(),
        "prediction": predictions,
        "answers": answers,
        "uid": uids,
        "question": questions,
    }
    return metrics
```

```python
@torch.no_grad()
def eval_loop_for_loss(model, dataloader, train_config, tokenizer):
    model.eval()
    losses = []
    for batch in tqdm.auto.tqdm(dataloader, desc="Evaluating", leave=False):
        batch = {k:v.to(train_config.device) for k,v in batch.items()}

        loss, _ = model(**batch)
        losses.append(loss.item())
    metrics = {
        "loss": sum(losses)/len(losses),
    }
    return metrics


@torch.no_grad()
def eval_loop_for_pretraining(model, dataloader, train_config, tokenizer):
    model.eval()

    losses = []
    ppls = []

    token_correct = 0
    token_total = 0

    for batch in tqdm.auto.tqdm(dataloader, desc="Evaluating", leave=False):
        batch = {k:v.to(train_config.device) for k,v in batch.items()}
        labels = batch.pop("labels")
        logits, _ = model(**batch)

        shifted_logits = logits[..., :-1, :].contiguous()
        shifted_labels = labels[..., 1:].contiguous()

        loss = torch.nn.functional.cross_entropy(shifted_logits.view(-1, shifted_logits.size(-1)), shifted_labels.view(-1),ignore_index=-100, reduction="none")

        loss = loss.view(shifted_labels.size())
        mask = shifted_labels.ne(-100)

        nll_loss = (loss * mask).sum() / mask.sum()
        ppl = torch.exp(nll_loss)

        losses.append(nll_loss.mean())
        ppls.append(ppl.mean())

        token_correct += (shifted_logits.argmax(-1) == shifted_labels).sum().item()
        token_total += shifted_labels.ne(-100).sum().item()

    metrics = {
        "loss": torch.mean(torch.stack(losses)).item(),
        "ppl": torch.mean(torch.stack(ppls)).item(),
        "token_acc": token_correct / token_total,
    }

    return metrics

def train(model, train_dataset, eval_dataset, train_collate_fn, eval_collate_fn, train_config, eval_loop):
    os.makedirs(train_config.logging_path, exist_ok=True)
    os.makedirs(train_config.output_path, exist_ok=True)

    train_model = model if isinstance(model, transformers.PreTrainedModel) else model.model
    if train_config.model_dtype == "cast_bf16":
        train_model = train_model.to(torch.bfloat16)

    print_model_statistics(train_model)

    train_model = train_model.to(train_config.device)
    train_model.train()
    train_model.compile()

    train_loader = torch.utils.data.DataLoader(
        train_dataset,
        batch_size=train_config.batch_size,
        shuffle=True,
        collate_fn=train_collate_fn,
    )
    eval_loader = torch.utils.data.DataLoader(
        eval_dataset,
        batch_size=train_config.eval_batch_size,
        shuffle=False,
        collate_fn=eval_collate_fn,
    )

    num_training_steps = len(train_loader) * train_config.num_train_epochs if train_config.max_steps is None else train_config.max_steps
    num_warmup_steps = train_config.warmup_steps
    num_train_epochs = train_config.num_train_epochs if train_config.max_steps is None else math.ceil(train_config.max_steps / len(train_loader))

    optimizer = torch.optim.AdamW(
        train_model.parameters(),
        lr=train_config.learning_rate,
        weight_decay=train_config.weight_decay,
    )
    scheduler = transformers.get_scheduler(
        train_config.lr_scheduler_type,
        optimizer=optimizer,
        num_warmup_steps=num_warmup_steps,
        num_training_steps=num_training_steps // train_config.gradient_accumulation_steps,
    )

    global_steps = 0

    logger = Logger(log_path=train_config.logging_path)
    with open(os.path.join(train_config.output_path, "train_config.json"), "w") as f:
        json.dump(train_config.to_dict(), f, indent=4)
    logger.log({"train_config": train_config.to_dict()})

    global_pbar = tqdm.auto.tqdm(total=num_training_steps, desc="Training")
    best_models = []
    loss_window = []
    for epoch in range(num_train_epochs):
        train_model.train()
        for batch in train_loader:
            batch = {k:v.to(train_config.device) for k,v in batch.items()}

            with torch.autocast(device_type=train_config.device, dtype=torch.float16, enabled=train_config.model_dtype == "amp_fp16"):
                loss, _ = train_model(**batch)
                l = loss.item()
                loss_window.append(l)
                if len(loss_window) > 100:
                    loss_window.pop(0)

            loss = loss / train_config.gradient_accumulation_steps
            loss.backward()

            if train_config.max_grad_norm is not None:
                torch.nn.utils.clip_grad_norm_(train_model.parameters(), train_config.max_grad_norm)

            if global_steps % train_config.gradient_accumulation_steps == 0:
                optimizer.step()
                scheduler.step()
                optimizer.zero_grad()
            global_steps += 1
            global_pbar.update(1)
            global_pbar.set_postfix({"epoch":epoch, "loss": f"{l:.4f}::{sum(loss_window)/len(loss_window):.4f}", "lr": f"{scheduler.get_last_lr()[0]:.2e}"})

            if global_steps % train_config.logging_interval == 0:
                logger.log({"steps": global_steps, "loss": l,"ave_loss":sum(loss_window)/len(loss_window), "lr": scheduler.get_last_lr()[0]})

            if global_steps % train_config.eval_interval == 0:
                global_pbar.disable = True
                global_pbar.refresh()
                eval_metrics = eval_loop(model, eval_loader, train_config, tokenizer)
                if "prediction" in eval_metrics: eval_metrics.pop("prediction")
                if "answers" in eval_metrics: eval_metrics.pop("answers")
                if "uid" in eval_metrics: eval_metrics.pop("uid")
                if "question" in eval_metrics: eval_metrics.pop("question")

                logger.log({"steps": global_steps, **eval_metrics},is_train=False)
                global_pbar.write(f"====== evaluation {global_steps} ====")
                for k, v in eval_metrics.items():
                    global_pbar.write(f"   {k}: {v}")
                global_pbar.write(f"======================")
                checkpoint = f"step-{global_steps}"

                best_models.append((checkpoint, eval_metrics[train_config.metric_for_best_model]))
                best_models.sort(key=lambda x: x[1], reverse=train_config.metric_greater_is_better)
                if len(best_models) > train_config.save_total_limit:
                    need_to_remove = best_models.pop()
                    if need_to_remove != checkpoint:
                        logger.log({"steps": global_steps, "remove_checkpoint": need_to_remove[0]})
                        shutil.rmtree(os.path.join(train_config.output_path, need_to_remove[0]), ignore_errors=True)
                        logger.log({"steps": global_steps, "save_chechkpoint": checkpoint})
                        os.makedirs(os.path.join(train_config.output_path, checkpoint), exist_ok=True)
                        train_model.save_pretrained(os.path.join(train_config.output_path, checkpoint))
                        tokenizer.save_pretrained(os.path.join(train_config.output_path, checkpoint))
                else:
                    logger.log({"steps": global_steps, "save_chechkpoint": checkpoint})
                    os.makedirs(os.path.join(train_config.output_path, checkpoint), exist_ok=True)
                    train_model.save_pretrained(os.path.join(train_config.output_path, checkpoint))
                    tokenizer.save_pretrained(os.path.join(train_config.output_path, checkpoint))
                global_pbar.disable = False
                global_pbar.refresh()
            if train_config.max_steps is not None and global_steps >= train_config.max_steps:
                break
        if train_config.max_steps is not None and global_steps >= train_config.max_steps:
            break

    eval_metrics = eval_loop(model, eval_loader, train_config, tokenizer)
    logger.log({"steps": global_steps, **eval_metrics},is_train=False)
    global_pbar.write(f"====== Training End ====")

    if "prediction" in eval_metrics: eval_metrics.pop("prediction")
    if "answers" in eval_metrics: eval_metrics.pop("answers")
    if "uid" in eval_metrics: eval_metrics.pop("uid")
    if "question" in eval_metrics: eval_metrics.pop("question")
    for k, v in eval_metrics.items():
        global_pbar.write(f"   {k}: {v}")
    global_pbar.write(f"======================")
    checkpoint = f"step-{global_steps}"

    best_models.append((checkpoint, eval_metrics[train_config.metric_for_best_model]))
    best_models.sort(key=lambda x: x[1], reverse=train_config.metric_greater_is_better)
    if best_models[0][0] != checkpoint:
        shutil.copytree(
            os.path.join(train_config.output_path, best_models[0][0]),
            os.path.join(train_config.output_path, "best_model"),
            dirs_exist_ok=True,
        )
    else:
        train_model.save_pretrained(os.path.join(train_config.output_path, "best_model"))
        tokenizer.save_pretrained(os.path.join(train_config.output_path, "best_model"))
```

```python
if DO_PROMPT_RAG:
    # Handout 4.2.1 (baseline RAG: chat template / few-shot / CoT / parsing)
    # and 4.3 (query-likelihood reranking, adaptive context assembly).
    from dataset.rag import donwload_dataset_nq_open_dpr, RAGDataset, RAGCollator
    from pyserini.search.lucene import LuceneSearcher
    from prompt_rag import PromptedRAG
    from model import TransformerForSequenceClassification
    from enhanced_rag import (UPRRerankRAG, AdaptiveAssemblyRAG, SlotOrderRAG,
                              GoldPositionRAG, OracleDIGRerankRAG)
    from improved_rag import ImprovedRAG, CrossEncoderRerankRAG
    from utils.metrics import best_subspan_exact_match, normalize_answer

    BM25_POOL_PATH = os.path.join(RESULTS_PATH, "bm25_top100.jsonl")
    # reranking scores are passage-scoped, so every configuration that reuses the
    # same pool and the same scorer pays for them once
    UPR_SCORE_CACHE = os.path.join(RESULTS_PATH, "upr_scores.jsonl")
    DIG_SCORE_CACHE = os.path.join(RESULTS_PATH, "oracle_dig_scores.jsonl")
    DIG_DUMP_PATH = os.path.join(RESULTS_PATH, "oracle_dig_dump.jsonl")
    # overridable from the environment so one notebook can serve every configuration
    PROMPT_RAG_NUM_SAMPLES = int(os.environ.get("PROMPT_RAG_NUM_SAMPLES", "0")) or None
    PROMPT_RAG_BATCH_SIZE = int(os.environ.get("PROMPT_RAG_BATCH_SIZE", "4"))
    PROMPT_RAG_TOPK = int(os.environ.get("PROMPT_RAG_TOPK", "5"))
    PROMPT_RAG_SUFFIX = os.environ.get("PROMPT_RAG_SUFFIX", "")

    PROMPT_RAG_CONFIGS = {
        # 4.2.1 -- naive prompt with controlled decoding, then the prompting variants
        "v0_control":  lambda: PromptedRAG(variant="v0"),
        "v1_zeroshot": lambda: PromptedRAG(variant="v1"),
        "v2_fewshot":  lambda: PromptedRAG(variant="v2"),
        "v3_json":     lambda: PromptedRAG(variant="v3"),
        "v4_cot":      lambda: PromptedRAG(variant="v4"),
        # 4.3 -- extensions on top of the baseline (v2)
        "ext_rerank":   lambda: UPRRerankRAG(variant="v2", k_pool=20, cache_path=BM25_POOL_PATH),
        "ext_assembly": lambda: AdaptiveAssemblyRAG(variant="v2", k_pool=20, cache_path=BM25_POOL_PATH,
                                                    rerank_enabled=False),
        "ext_full":     lambda: AdaptiveAssemblyRAG(variant="v2", k_pool=20, cache_path=BM25_POOL_PATH),
        # v3_json was the strongest prompt and reranking was the strongest retrieval change,
        # so the proposed system combines them; k_pool=50 tests the larger oracle the
        # diagnostics exposed (hasanswer@20 = 0.717 vs hasanswer@50 = 0.801)
        "ext_rerank_v3":     lambda: UPRRerankRAG(variant="v3", k_pool=20, cache_path=BM25_POOL_PATH),
        "ext_rerank_v3_p50": lambda: UPRRerankRAG(variant="v3", k_pool=50, cache_path=BM25_POOL_PATH),
        "ext_rerank_v3_p100": lambda: UPRRerankRAG(variant="v3", k_pool=100, cache_path=BM25_POOL_PATH),
        # position study: identical retrieval to ext_rerank_v3_p50, only the slot
        # each passage occupies in the prompt changes
        "slot_edge": lambda: SlotOrderRAG(variant="v3", k_pool=50, slot_order="edge",
                                          cache_path=BM25_POOL_PATH,
                                          rerank_cache_path=UPR_SCORE_CACHE),
        "slot_edge_last": lambda: SlotOrderRAG(variant="v3", k_pool=50, slot_order="edge_last",
                                               cache_path=BM25_POOL_PATH,
                                               rerank_cache_path=UPR_SCORE_CACHE),
        "slot_asc": lambda: SlotOrderRAG(variant="v3", k_pool=50, slot_order="asc",
                                         cache_path=BM25_POOL_PATH,
                                         rerank_cache_path=UPR_SCORE_CACHE),
        # oracle ceiling for answer-likelihood reranking (InfoGain-RAG); uses the
        # gold answer, so it bounds what a distilled reranker could reach and is
        # never reported as a method
        "oracle_dig": lambda: OracleDIGRerankRAG(variant="v3", k_pool=50,
                                                 cache_path=BM25_POOL_PATH,
                                                 rerank_cache_path=DIG_SCORE_CACHE,
                                                 dev_path=NQ_DEV_PATH,
                                                 dig_dump_path=DIG_DUMP_PATH),
    }
    # --- the three borrowed ideas, all on top of the reranked k_pool=50 pool ---
    # EXIT: sentences instead of passages, inside the measured 731-token budget of
    # the top-5 context they replace, so coverage rises at constant prompt size.
    # InfoGain: fuse UPR with the likelihood of a first-pass answer.
    # GRAD: put the best evidence last, against the measured position bias.
    IMPROVED_COMMON = dict(variant="v3", k_pool=50, cache_path=BM25_POOL_PATH,
                           rerank_cache_path=UPR_SCORE_CACHE)
    SENT_CACHE = os.path.join(RESULTS_PATH, "sentence_scores.jsonl")
    PROXY_CACHE = os.path.join(RESULTS_PATH, "proxy_dig_scores.jsonl")
    FIRST_PASS = os.path.join(RESULTS_PATH, "prompt_rag_slot_edge_last_output.txt")
    PROMPT_RAG_CONFIGS.update({
        "sent20": lambda: ImprovedRAG(compress_from=20, sentence_cache_path=SENT_CACHE,
                                      **IMPROVED_COMMON),
        "sent20_last": lambda: ImprovedRAG(compress_from=20, slot_order="edge_last",
                                           sentence_cache_path=SENT_CACHE, **IMPROVED_COMMON),
        "sent50": lambda: ImprovedRAG(compress_from=50, slot_order="edge_last",
                                      sentence_cache_path=SENT_CACHE, **IMPROVED_COMMON),
        "dig_proxy": lambda: ImprovedRAG(use_proxy=True, pseudo_path=FIRST_PASS,
                                         proxy_cache_path=PROXY_CACHE, **IMPROVED_COMMON),
        "dig_order": lambda: ImprovedRAG(use_proxy=True, pseudo_path=FIRST_PASS,
                                         proxy_cache_path=PROXY_CACHE,
                                         slot_order="edge_last", **IMPROVED_COMMON),
        "sent20_dig": lambda: ImprovedRAG(compress_from=20, sentence_cache_path=SENT_CACHE,
                                          use_proxy=True, pseudo_path=FIRST_PASS,
                                          proxy_cache_path=PROXY_CACHE, **IMPROVED_COMMON),
        "all_three": lambda: ImprovedRAG(compress_from=20, sentence_cache_path=SENT_CACHE,
                                         use_proxy=True, pseudo_path=FIRST_PASS,
                                         proxy_cache_path=PROXY_CACHE,
                                         slot_order="edge_last", **IMPROVED_COMMON),
        # does the answer-likelihood signal work on its own, or only fused?
        "dig_only": lambda: ImprovedRAG(use_proxy=True, fuse=False, pseudo_path=FIRST_PASS,
                                        proxy_cache_path=PROXY_CACHE,
                                        slot_order="edge_last", **IMPROVED_COMMON),
        # a 63M scorer costs about a twentieth of the 1.2B one per passage; reranking
        # only needs an ordering, so the scorer need not share the reader's vocabulary.
        # three checkpoints answer three different questions: whether a plain LM's query
        # likelihood carries relevance at this scale, whether task finetuning helps or
        # erodes it, and whether an untrained network scores just as well -- which would
        # mean the signal is an artefact
        "gptsmall_rerank": lambda: _gpt_small_scorer("pretraining"),
        "gptsmall_rag_rerank": lambda: _gpt_small_scorer("rag"),
        "gptsmall_random_rerank": lambda: _gpt_small_scorer("pretraining_random"),
        # the trained reranker: GPT-small distilled from DIG labels computed on the
        # NQ train split, scoring as a cross-encoder rather than by query likelihood
        "trained_rerank": lambda: _trained_reranker(),
        # the controlled sweep found monotone recency with no primacy, so mapping rank
        # to slot monotonically is the arrangement the measurement motivates; edge_last
        # is a leftover of the lost-in-the-middle reading it replaced
        "dig_asc": lambda: ImprovedRAG(use_proxy=True, pseudo_path=FIRST_PASS,
                                       proxy_cache_path=PROXY_CACHE,
                                       slot_order="asc", **IMPROVED_COMMON),
        # the k_pool question itself: does a better ranker let a shallower pool do
        # the job of a deeper one? scores for the top-20 are a subset of the cached
        # top-50, so this costs generation only
        "dig_p20": lambda: ImprovedRAG(use_proxy=True, pseudo_path=FIRST_PASS,
                                       proxy_cache_path=PROXY_CACHE, slot_order="edge_last",
                                       variant="v3", k_pool=20, cache_path=BM25_POOL_PATH,
                                       rerank_cache_path=UPR_SCORE_CACHE),
    })
    # RFT reader: the dig_order pipeline unchanged, but the reader names the
    # passage it used so the citation terms in the reward are measurable. The
    # retrieval side is identical, so HasAnswer must not move between these and
    # dig_order; adapter=None is the untrained control for the contract change.
    from reader_rft_rag import SourceReaderRAG

    def _rft(adapter=None):
        return SourceReaderRAG(adapter_path=adapter, use_proxy=True,
                               pseudo_path=FIRST_PASS, proxy_cache_path=PROXY_CACHE,
                               slot_order="edge_last", **IMPROVED_COMMON)

    PROMPT_RAG_CONFIGS.update({
        "rft_g1": lambda: _rft(None),
        "rft_c1": lambda: _rft(os.path.join("output", "reader_rft", "c1")),
        "rft_r1": lambda: _rft(os.path.join("output", "reader_rft", "r1")),
        "rft_r2": lambda: _rft(os.path.join("output", "reader_rft", "r2")),
        "rft_r3": lambda: _rft(os.path.join("output", "reader_rft", "r3")),
    })
    # --- selection experiments -------------------------------------------
    # The reader is the stock v3 reader with frozen weights; only which passages
    # reach it changes. SELECTOR_LOG keeps the per-question choices so B and C
    # can be compared pairwise afterwards.
    from selector_rag import SelectedPassageRAG
    from multi_select_rag import MultiSelectRAG

    SELECTOR_DIR = os.path.join("output", "selector")
    MULTI_DIR = os.path.join("output", "multi_select")

    def _sel(mode="selector", adapter=None):
        return SelectedPassageRAG(mode=mode, adapter_path=adapter,
                                  dev_path=NQ_DEV_PATH, use_proxy=True,
                                  pseudo_path=FIRST_PASS, proxy_cache_path=PROXY_CACHE,
                                  slot_order="edge_last", **IMPROVED_COMMON)

    def _msel(name, mode="selector", adapter=None):
        return MultiSelectRAG(mode=mode, adapter_path=adapter, dev_path=NQ_DEV_PATH,
                              log_path=os.path.join(MULTI_DIR, "%s_selection.jsonl" % name),
                              use_proxy=True, pseudo_path=FIRST_PASS,
                              proxy_cache_path=PROXY_CACHE, slot_order="edge_last",
                              **IMPROVED_COMMON)

    PROMPT_RAG_CONFIGS.update({
        # single selection: one passage or none
        "sel_oracle1": lambda: _sel("oracle"),
        "sel_abstain": lambda: _sel("abstain"),
        "sel_b": lambda: _sel("selector", None),
        "sel_c": lambda: _sel("selector", os.path.join(SELECTOR_DIR, "grpo_r1")),
        # multi selection: any subset, including the empty one
        "msel_a": lambda: _msel("msel_a", "all"),
        "msel_b": lambda: _msel("msel_b", "selector", None),
        "msel_c": lambda: _msel("msel_c", "selector",
                                os.path.join(MULTI_DIR, "grpo_r1")),
        "msel_d": lambda: _msel("msel_d", "empty"),
        "msel_e": lambda: _msel("msel_e", "slot5"),
        "msel_oracle": lambda: _msel("msel_oracle", "oracle"),
    })
    # both axes at once: the GRPO selector in front of the RFT-trained reader.
    # Neither prior run had the other's adapter, so this is the combination the
    # results do not cover. Separate model instances, not named adapters.
    from combined_rag import CombinedRAG

    PROMPT_RAG_CONFIGS.update({
        "comb_c": lambda: CombinedRAG(
            selector_adapter=os.path.join(MULTI_DIR, "grpo_r1"),
            adapter_path=os.path.join("output", "reader_rft", "r3"),
            dev_path=NQ_DEV_PATH,
            log_path=os.path.join(MULTI_DIR, "comb_c_selection.jsonl"),
            use_proxy=True, pseudo_path=FIRST_PASS, proxy_cache_path=PROXY_CACHE,
            slot_order="edge_last", **IMPROVED_COMMON),
        # control: same trained reader, no selection (= rft_r3 re-expressed)
        "comb_all": lambda: CombinedRAG(
            selector_adapter=None,
            adapter_path=os.path.join("output", "reader_rft", "r3"),
            dev_path=NQ_DEV_PATH,
            log_path=os.path.join(MULTI_DIR, "comb_all_selection.jsonl"),
            use_proxy=True, pseudo_path=FIRST_PASS, proxy_cache_path=PROXY_CACHE,
            slot_order="edge_last", **IMPROVED_COMMON),
    })
    # controlled position sweep: the answer-bearing passage is pinned to one slot
    # and the rest are answer-free, so position is the only variable
    PROMPT_RAG_CONFIGS.update({
        f"goldpos{slot}": (lambda s=slot: GoldPositionRAG(
            variant="v3", k_pool=50, gold_slot=s, cache_path=BM25_POOL_PATH,
            rerank_cache_path=UPR_SCORE_CACHE, dev_path=NQ_DEV_PATH))
        for slot in range(5)
    })
    PROMPT_RAG_SELECTED = [name.strip() for name in
                           os.environ.get("PROMPT_RAG_SELECTED", "v2_fewshot").split(",") if name.strip()]

    rag_cache_path = os.path.join(LOCAL_CACHE_PATH, "rag")
    donwload_dataset_nq_open_dpr(rag_cache_path)
    NQ_DEV_PATH = os.path.join(rag_cache_path, "data", "nq_open_dpr", "nq_dev")

    GPTSMALL_SCORERS = {}

    def _gpt_small_scorer(checkpoint):
        """ImprovedRAG whose reranking is done by GPT-small instead of Llama."""
        if checkpoint not in GPTSMALL_SCORERS:
            path = os.path.join(OUTPUT_PATH, checkpoint, "best_model")
            small = TransformerForCausalLM.from_pretrained(path).to("cuda").eval()
            # make_random_init.py writes weights only; every GPT-small checkpoint
            # shares the tokenizer the pretraining run saved
            tok_path = path if os.path.exists(os.path.join(path, "tokenizer.json"))                 else os.path.join(OUTPUT_PATH, "pretraining", "best_model")
            small_tok = transformers.AutoTokenizer.from_pretrained(tok_path)
            small_tok.padding_side = "left"
            if small_tok.pad_token is None:
                small_tok.pad_token = small_tok.eos_token
            GPTSMALL_SCORERS[checkpoint] = (small, small_tok)
        model, tok = GPTSMALL_SCORERS[checkpoint]
        rag = ImprovedRAG(variant="v3", k_pool=50, slot_order="edge_last",
                          cache_path=BM25_POOL_PATH,
                          rerank_cache_path=os.path.join(
                              RESULTS_PATH, "upr_scores_gptsmall_%s.jsonl" % checkpoint),
                          score_max_passage_tokens=160, passages_per_score_batch=50)
        return rag.set_scorer(model, tok, "gptsmall-" + checkpoint)

    def _trained_reranker():
        path = os.path.join(OUTPUT_PATH, "reranker")
        cross = TransformerForSequenceClassification.from_pretrained(
            path, num_labels=2).to("cuda").eval()
        cross_tok = transformers.AutoTokenizer.from_pretrained(path)
        cross_tok.padding_side = "left"
        if cross_tok.pad_token is None:
            cross_tok.pad_token = cross_tok.eos_token
        rag = CrossEncoderRerankRAG(variant="v3", k_pool=50, slot_order="edge_last",
                                    cache_path=BM25_POOL_PATH,
                                    rerank_cache_path=os.path.join(
                                        RESULTS_PATH, "cross_encoder_scores.jsonl"),
                                    passages_per_score_batch=50)
        return rag.set_scorer(cross, cross_tok, "gptsmall-dig-reranker")

    PREBUILT_INDEX_NAME_BM25 = "wikipedia-dpr-100w"
    LOCAL_INDEX_NAME_BM25 = os.path.join(rag_cache_path, "lucene-index.wikipedia-dpr-100w.20210120.d1b9e6")
    try:
        searcher_bm25 = LuceneSearcher(index_dir=LOCAL_INDEX_NAME_BM25)
    except:
        searcher_bm25 = LuceneSearcher.from_prebuilt_index(prebuilt_index_name=PREBUILT_INDEX_NAME_BM25)
        index_dir = searcher_bm25.index_dir
        shutil.move(index_dir, LOCAL_INDEX_NAME_BM25)
        searcher_bm25 = LuceneSearcher(index_dir=LOCAL_INDEX_NAME_BM25)

    model_name_or_path = "meta-llama/Llama-3.2-1B-Instruct"
    llm_model = transformers.AutoModelForCausalLM.from_pretrained(model_name_or_path,
                                                                  torch_dtype=torch.bfloat16,
                                                                  device_map={"": 0},
                                                                  low_cpu_mem_usage=True,)
    llm_tokenizer = transformers.AutoTokenizer.from_pretrained(model_name_or_path)
    llm_tokenizer.pad_token = llm_tokenizer.eos_token
    llm_tokenizer.padding_side = "left"

    eval_dataset = RAGDataset(
        tokenizer=llm_tokenizer,
        is_train=False,
        dataset_path=os.path.join(rag_cache_path, "data", "nq_open_dpr", "nq_dev"),
        num_samples=PROMPT_RAG_NUM_SAMPLES,
    )
    eval_collator = RAGCollator(tokenizer=llm_tokenizer, is_train=False)
    eval_loader = torch.utils.data.DataLoader(
        eval_dataset,
        batch_size=PROMPT_RAG_BATCH_SIZE,
        shuffle=False,
        collate_fn=eval_collator,
    )

    def _has_answer(passages, answers):
        answers_norm = [normalize_answer(a) for a in answers]
        for passage in passages:
            text = normalize_answer(f"{passage.get('title', '')} {passage.get('text', '')}")
            if any(a and a in text for a in answers_norm):
                return True
        return False

    @torch.no_grad()
    def eval_for_prompted_rag(model_rag, dataloader, k=5):
        model_rag.model.eval()

        model_rag.reset_stats()
        uids, questions, raws, answers = [], [], [], []
        retrieved_hit, context_hit = 0, 0
        for batch in tqdm.auto.tqdm(dataloader, desc="Evaluating", leave=False):
            outputs = model_rag.retrieval_augmented_generate(
                queries=batch["question"], qids=batch["uid"], k=k)
            raw = model_rag.tokenizer.batch_decode(outputs, skip_special_tokens=True)

            # retrieved = what search returned, context = what actually entered the prompt;
            # AdaptiveAssemblyRAG drops passages, so the two differ and must not be conflated
            for passages, answer in zip(model_rag.last_retrieved, batch["answers"]):
                retrieved_hit += int(_has_answer(passages, answer))
            for passages, answer in zip(model_rag.last_contexts, batch["answers"]):
                context_hit += int(_has_answer(passages, answer))

            uids.extend(batch["uid"])
            questions.extend(batch["question"])
            raws.extend(raw)
            answers.extend(batch["answers"])

        predictions = model_rag.postprocess(raws)
        accuracy = best_subspan_exact_match(predictions, answers)
        rouge_score = rouge.compute(predictions=predictions, references=answers)
        metrics = {
            "accuracy": accuracy["acc"],
            "rouge1": rouge_score["rouge1"].item(),
            "rouge2": rouge_score["rouge2"].item(),
            "rougeL": rouge_score["rougeL"].item(),
            "rougeLsum": rouge_score["rougeLsum"].item(),
            f"hasanswer_retrieved@{k}": retrieved_hit / len(uids),
            "hasanswer_in_prompt": context_hit / len(uids),
            "mean_pred_words": model_rag.parse_stats.get("mean_pred_words"),
            "pct_empty": model_rag.parse_stats.get("pct_empty"),
            "parse_stats": model_rag.parse_stats,
            "assembly_stats": getattr(model_rag, "assembly_stats", {}),
            "n": len(uids),
        }
        return metrics, uids, questions, predictions, raws, answers

    all_scores = {}
    for config_name in PROMPT_RAG_SELECTED:
        model_rag = PROMPT_RAG_CONFIGS[config_name]()
        model_rag.set_model(llm_model)
        model_rag.set_tokenizer(llm_tokenizer)
        model_rag.set_retriever(searcher_bm25)

        metrics, uid, questions, prediction, raw, answers = eval_for_prompted_rag(
            model_rag, eval_loader, k=PROMPT_RAG_TOPK)

        print(f"====== evaluation: {config_name} ====")
        for key, value in metrics.items():
            print(f"   {key}: {value}")
        print(f"======================")

        all_scores[config_name] = metrics
        with open(os.path.join(RESULTS_PATH, f"prompt_rag_{config_name}{PROMPT_RAG_SUFFIX}_score.json"), "w") as f:
            json.dump(metrics, f, indent=4)
        with open(os.path.join(RESULTS_PATH, f"prompt_rag_{config_name}{PROMPT_RAG_SUFFIX}_output.txt"), "w", encoding="utf-8") as f:
            for u, q, p, a, r in zip(uid, questions, prediction, answers, raw):
                f.write(f"{u}\t{q}\t{a}\t{p}\t{r.replace(chr(10), ' ')}\n")

        del model_rag
        torch.cuda.empty_cache()

    with open(os.path.join(RESULTS_PATH, f"prompt_rag_summary{PROMPT_RAG_SUFFIX}.json"), "w") as f:
        json.dump(all_scores, f, indent=4)

    del llm_model, eval_dataset
    torch.cuda.empty_cache()
```
